```python
import sys
assert sys.version_info[:2] == (3, 12), 'Fork the working Python 3.12 notebook and pin its original environment'
print('Python runtime:', sys.version)
import base64, gzip, json, os
from pathlib import Path
root = Path('/kaggle/working/patchquestions-code')
root.mkdir(exist_ok=True)
files = json.loads(gzip.decompress(base64.b64decode('')))
for name, encoded in files.items():
    (root/name).write_bytes(base64.b64decode(encoded))
print('Extracted', len(files), 'versioned files')
```

```python
import subprocess, sys
subprocess.run([sys.executable, str(root/'offline_install.py')], check=True)
```

```python
with (root/'worker.log').open('w') as log:
    process = subprocess.Popen([sys.executable, '-u', str(root/'gpu_worker.py')], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    for line in process.stdout:
        log.write(line); log.flush(); print(line, end='', flush=True)
    returncode = process.wait()
if returncode:
    raise subprocess.CalledProcessError(returncode, process.args)
```
